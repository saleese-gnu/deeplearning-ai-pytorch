✅ (2) OpenWeather API 직접 호출 버전 (추천)

이 버전은 직접 만든 LangChain Tool을 통해 실시간 날씨를 가져옵니다.
(OpenWeather 무료 키: https://openweathermap.org/api
 에서 발급 가능)

```python
import os
from dotenv import load_dotenv

# 바로 임포트
# os.environ["OPENAI_API_KEY"] ="sk-proj-IeiPd3N4ZP..."

# .env 파일 로드
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
print(api_key)
OPENWEATHER_API_KEY = os.getenv("OPENWEATHER_API_KEY")
```

```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_openai import ChatOpenAI
import requests, os

@tool("SeoulWeather")
def get_weather_in_seoul(query: str) -> str:
    """서울의 현재 날씨를 알려주는 도구입니다."""
    url = f"http://api.openweathermap.org/data/2.5/weather?q=Seoul&appid={OPENWEATHER_API_KEY}&lang=kr&units=metric"
    response = requests.get(url).json()
    desc = response["weather"][0]["description"]
    temp = response["main"]["temp"]
    return f"현재 서울의 날씨는 {desc}, 온도는 {temp}°C 입니다."

agent = create_agent(
    tools=[get_weather_in_seoul],
    model="gpt-4o-mini"
)

# Run the agent
response = agent.invoke(
    {"messages": [{"role": "user", "content": "오늘 서울 날씨를 알려줘"}]}
)

print(response["messages"][-1].content)
```
