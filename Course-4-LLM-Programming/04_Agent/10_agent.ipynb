```python
# !pip install -U langchain

from langchain.agents import create_agent
from langchain_experimental.tools.python.tool import PythonREPLTool

# PythonREPL 도구 생성
python_repl = PythonREPLTool()

# 에이전트 생성
agent = create_agent(
    model="gpt-4o-mini",
    tools=[python_repl],
    system_prompt=(
        "You are a helpful assistant that can execute Python code using the PythonREPL tool. "
        "Use it for math, plotting, or quick computations."
    ),
)

# 사용자 요청 실행
response = agent.invoke(
    {
        "messages": [
            {"role": "user", "content": "Calculate the sum of squares of numbers from 1 to 10."}
        ]
    }
)

# 결과 출력
print(response["messages"][-1].content)
```
