```python
from langchain.agents import create_agent
from langchain.tools import tool
from ddgs import DDGS

# Define the DuckDuckGo search tool
@tool("duckduckgo_search", description="Searches the web using DuckDuckGo and returns concise results.")
def duck_search(query: str) -> str:
    """Perform a web search using DuckDuckGo and return short snippets."""
    try:
        with DDGS() as ddgs:
            results = ddgs.text(query, max_results=3)
            formatted = "\n".join(
                [f"- {r['title']}: {r['body']} ({r['href']})" for r in results]
            )
            return formatted if formatted else "No results found."
    except Exception as e:
        return f"Search error: {e}"

# Create the agent
agent = create_agent(
    model="gpt-4o-mini",
    tools=[duck_search],
    system_prompt="You are a helpful web assistant that searches real-time information using the duckduckgo_search tool.",
)

# Run the agent
response = agent.invoke(
    {"messages": [{"role": "user", "content": "Search for recent advancements in quantum computing."}]}
)

# Print the assistant's reply
print(response["messages"][-1].content)
```
