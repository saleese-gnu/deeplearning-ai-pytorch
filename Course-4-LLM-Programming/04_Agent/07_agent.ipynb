```python
# !pip install -U langchain numexpr

from langchain.agents import create_agent
from langchain.tools import tool
import numexpr as ne

# Define the llm-math tool using numexpr
@tool("llm-math", description="Evaluates mathematical expressions safely and efficiently using numexpr.")
def llm_math(expression: str) -> str:
    """Safely evaluate math expressions using numexpr."""
    try:
        result = ne.evaluate(expression)
        return str(result.item() if hasattr(result, "item") else result)
    except Exception as e:
        return f"Error evaluating expression: {e}"

# Create the agent
agent = create_agent(
    model="gpt-4o-mini",
    tools=[llm_math],
    system_prompt="You are a helpful assistant that performs math calculations using the llm-math tool.",
)

# Run the agent
response = agent.invoke(
    {"messages": [{"role": "user", "content": "What is (3.5 + 4.2) * 2?"}]}
)

# Print the assistant's reply
print(response["messages"][-1].content)
```
