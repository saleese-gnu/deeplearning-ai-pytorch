```python
from langchain.agents import create_agent
from langchain.tools import tool
import wikipedia

# Define the Wikipedia search tool
@tool("wikipedia_search", description="Searches Wikipedia and returns a summary of a topic.")
def wiki_search(query: str) -> str:
    """Search for a topic on Wikipedia and return the summary."""
    try:
        return wikipedia.summary(query, sentences=3)
    except wikipedia.DisambiguationError as e:
        return f"Multiple results found: {e.options[:5]}"
    except wikipedia.PageError:
        return "No page found for this query."

# Create the agent
agent = create_agent(
    model="gpt-4o-mini",
    tools=[wiki_search],
    system_prompt="You are a helpful assistant that can retrieve information from Wikipedia using the wikipedia_search tool.",
)

# Run the agent
response = agent.invoke(
    {"messages": [{"role": "user", "content": "Tell me about the history of the Internet."}]}
)

# Print the assistant's reply
print(response["messages"][-1].content)
```
