```python
from langchain.agents import create_agent

def convert_currency(amount: float, from_currency: str, to_currency: str) -> str:
    """Convert currency from one unit to another (mock example)."""
    # 실제 환율 계산 대신 단순히 예시 비율 사용
    rates = {
        ("USD", "KRW"): 1350,
        ("KRW", "USD"): 1 / 1350,
        ("USD", "EUR"): 0.9,
        ("EUR", "USD"): 1.1,
    }

    rate = rates.get((from_currency.upper(), to_currency.upper()))
    if not rate:
        return f"Sorry, I can't convert from {from_currency} to {to_currency}."

    converted = amount * rate
    return f"{amount:.2f} {from_currency.upper()} = {converted:.2f} {to_currency.upper()}"

# Agent 생성
agent = create_agent(
    model="gpt-4o-mini",
    tools=[convert_currency],
    system_prompt="You are a financial assistant that can convert currencies.",
)

# Agent 실행
response = agent.invoke({
    "messages": [
        {"role": "user", "content": "convert 100 USD to KRW"}
    ]
})

print(response["messages"][2].content)
```
